## Markdown

Markdown is a special text formatting language. The overview:

* **Headings:** `# H1`, `## H2`, `### H3`
* **Emphasis:** `*italic*`, `**bold**`, `~~strikethrough~~`.
* **Lists:**

  * Unordered: `- item` or `* item`
  * Ordered: `1. item`

* **Links:** `[text](https://example.com)`
* **Images:** `![alt text](image.png)`
* **Code:**
  * Inline: `` `code` ``
  * Block:

    ````markdown
    ```python
    print("hello")
    ```
    ````

* **Blockquotes:** `> quote`
* **Tables:**
  ```markdown
  | A | B |
  |:-|:-:|
  | 1 | 2 |
  ```

* **Horizontal line:** `---` or `***`
* **Inline HTML:** `<b>bold</b>`

It’s designed to be readable as plain text and easily converted to HTML.

There are many implementations of Mardown; in fact, each big enough tool that utilises markdow implements its own dialect. However, the default standard to refer is the [CommonMark Spec](https://spec.commonmark.org/).

### Tables

Tables in markdown can be specified with the following syntax:

```markdown
| Column name 1 | Column name 2 | Column name 3 |
|:--------------|--------------:|:-------------:|
|   value 11    |  value 12     |  value 13     |                
|   value 21    |  vlaue 22     |  value 23     |
```

The colon (`:`) in the line separating the header from the content defines content's alignment.

The example table will be rendered as follows:

| Column name 1 | Column name 2 | Column name 3 |
|:--------------|--------------:|:-------------:|
|   value 11    |  value 12     |  value 13     |                
|   value 21    |  vlaue 22     |  value 23     |

### Links

- Any text that begins with `https://` will be interpreted as a link and will be clickable.
- To create a hyperlink use the syntax `[represented text](<addres>)`.
- It's not alway convenient to put the address right after text. You can define the hyperlink as `[my link][1]` and specify the address later in the text as `[1]: <address>`. This approach is called **refence-style link**.

## JSON

JSON (JavaScript Object Notation) is a lightweight, text-based data interchange format designed to be easy for both humans to read and machines to parse. It represents data using objects (key–value pairs), arrays (ordered lists), strings, numbers, booleans, and the `null` value. JSON has a simple, language-independent syntax originally derived from JavaScript object literals, but it is supported by virtually every modern programming language. The format is formally standardized in **ECMA-404** (syntax) and **RFC 8259** (Internet interoperability and recommended practices). JSON is widely used for configuration files, REST APIs, data serialization, and communication between distributed systems.

There are a lot of tooling for JSON, but for me important were:

- A JSON schema is a special format based on a JSON itself. It allows to declare the format of json object, so JSON documents could be validated against schemas.
- A `jq` is command-line utility that parses JSON documents. Knowing the basics of this tool is extremely useful, as it allows you to immediately fetch useful information from JSON documents, which sometimes can be extremely large.

### JSON schema

JSON schema is a declarative protocol to specify the format of the JSON documents. 

For more check:

- The [Official website](https://json-schema.org/).
- For implemetations of JSON schema visit [tools](https://json-schema.org/tools?query=&sortBy=name&sortOrder=ascending&groupBy=toolingTypes&licenses=&languages=&drafts=&toolingTypes=&environments=&showObsolete=false&supportsBowtie=false):
    - The python implementation is [jsonschema](https://github.com/python-jsonschema/jsonschema) package.
    - The python package with CLI [check-jsonschema](https://pypi.org/project/check-jsonschema/).
- The [JSON Schema](formats/json_schema.ipynb) page.

---

The following cell specifies a schema in JSON format with the attributes "price" and "name". The "price" attribute must be an integer value, and "name" is declared to be a string.

`````bash
cat << EOF > /tmp/schema.json 
{
    "type" : "object",
    "properties" : {
        "price" : {"type" : "number"},
        "name" : {"type" : "string"}
    }
}
EOF
`````

The `check-jsonschema` verifies that the files passed correspond to the schema specified by the `--schemafile` argument.

`````bash
cat << EOF > /tmp/object.json
{"name": "Eggs", "price": 43.99}
EOF
check-jsonschema --schemafile /tmp/schema.json /tmp/object.json
`````

Validation is successful. However, the attempt to pass a number as a string to the `price` in the following cell fails.

`````bash
cat << EOF > /tmp/object.json
{"name" : "Eggs", "price" : "34.99"}
EOF
check-jsonschema --schemafile /tmp/schema.json /tmp/object.json
`````

## YAML

YAML or YML is human-readable computer format. The format provides simple, minimalistic syntax that is easy for humans to read. It's main niche is configuration files for systems.

The main features of the format:

- **Hierarchical structure** - Naturally represents nested maps and lists.
- **Key-value mappings** - Supports dictionaries/objects.
- **Sequences (lists)** - Uses - for ordered collections.
- **Scalars** - Supports strings, numbers, booleans, null, dates, and more.
- **Comments** - `#` introduces comments.
- **Multiple documents** - a single file can contain multiple documents separated by ---.
- **Anchors and aliases** - reuse data with `&`anchor and `*`alias.
- **Merge keys** - Combine mappings with `<<`.
- **Block and flow styles** - Collections can be written in multiline or JSON-like inline syntax.
- **Multiline strings** - Literal (`|`) and folded (`>`) block scalars.
- **Explicit typing** (optional) - Type tags such as `!!str` or custom tags.
- **JSON superset** - Valid JSON is also valid YAML (YAML 1.2).
- **Extensible** - Applications can define custom tags and schemas.

Check the [YAML Specification Index](https://yaml.org/spec/) for official reference to the format.

## DateTime

There are two main groups to represent a date in a computer:

- **Positional**: separate numbers for each unit of time (year, month, day, hour, second).
- **Timestamp**: This represents the date as an integer number of units passed from a point in time. It typically represents the number of microseconds passed since 1 January 1970, but other options for the unit and start point are possible.

**Time zones** are typically represented as +/-number of hours in UTC. **UTC** is (Coordinated Universal Time) corresponds to the time in GMT, and is typically used in computer systems to measure time in order to avoid issues relating to timezones.

**ISO-8601** is a most popular way to write some date as a text. It's a bit funny that there is no open publication of the standard, you have to pay for it. Check the corresponding [ISO page](https://www.iso.org/iso-8601-date-and-time-format.html).

The most important of the ISO-8601:

- Date have format `yyyy-mm-dd`, for example `1991-12-08`. 
- Time have format `hh:mm:ss`, for example `24:55:43`.
- The datetime joins date and time reprsentation though `T` symbol - `yyyy-mm-ddThh:mm:ss`, for example `2000-10-05T16:10:22`.
- To specify the time zone for a time value, use a time zone offset after the time in the format `[+-]HH:MM`. The offset indicates how much time should be added to or subtracted from UTC to obtain the local time. For example, datetime `1941-06-22T04:00:00+03:00` specifies time in Belarus.
